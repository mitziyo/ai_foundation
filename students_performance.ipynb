```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("student_performance.csv")

print("Dataset shape (rows, columns):", df.shape)
print("Column name:", df.columns.tolist())

print("First 5 rows of the dataset")
print(df.head())

print("Summary statistics:")
print(df.describe())

print("Data types and null info:")
print(df.info())
```

```python
print("Average score by columns:")
df[["total_score", "weekly_self_study_hours"]].mean()
```

```python
print("Maxumum scores by participation")
df[["class_participation", "attendance_percentage"]].max()
```

```python
print("Minimum scores by participation")
df[["class_participation", "attendance_percentage"]].min()
```

```python
# Correlation ikki ozgaruvchi orasidagi bogliqlik
print("Correlation:")
df[["class_participation", "attendance_percentage"]].corr()
```

```python
df["total score"] = df[["class_participation", "attendance_percentage"]].sum(axis=1)
df["average score"] = df["total score"] / 2
df["passed"] = np.where(df["average score"] >= 50, "Yes", "No")

print("New columns preview (total, average, passed):")
print(df[["total score", "average score", "passed"]].head())
```

```python
grade_group = df.groupby("grade")[["class_participation", "weekly_self_study_hours", "attendance_percentage"]].mean()
print("student score by test preperation course:")
print(grade_group)
print(df.columns.tolist())

# Who passed vs failed
pass_rate = df["passed"].value_counts(normalize=True) * 100
print("Pass rate:\n", pass_rate)

```

```python
# Convert scores to Numpy array
scores = df[["class_participation", "weekly_self_study_hours", "attendance_percentage"]].values

# Mean, Std Deviation
print("Numpy Mean:", np.mean(scores, axis=0))
print("Numpy Std:", np.std(scores, axis=0))

# Z-score (normalize)
z_score = (scores - np.mean(scores, axis=0)) / np.std(scores, axis=0)
print("Z-scores (normalized numbers)")
print(z_score[:5])

# Average 0
# Negative -
# Positive +
```

```python
# How many students scored above 90 average?
top_scorers = df[df["average score"] >= 57]
print("Top scored student above 90:", len(top_scorers))

 # How many failed?
failed = df[df["passed"] == "No"]
print("Failed from exam:", len(failed))
```
